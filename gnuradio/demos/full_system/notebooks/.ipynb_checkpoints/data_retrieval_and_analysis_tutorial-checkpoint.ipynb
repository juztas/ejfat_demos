# Data Retrieval and Analysis Tutorial

**Interactive guide for retrieving and analyzing Bluesky experiment data**

This notebook demonstrates how to:
- Access experiment data from DataBroker
- Perform statistical analysis
- Create visualizations
- Export data in multiple formats

---

## Setup

First, let's import all necessary modules and set up our environment.

```python
# Standard imports
import sys
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from datetime import datetime

# Add parent directory to path
sys.path.insert(0, str(Path.cwd().parent))

# Bluesky imports
from databroker import Broker

# Analysis imports
from analysis.statistics import (
    analyze_run, find_peaks, compare_runs,
    compute_correlation, fit_polynomial,
    detect_outliers, compute_snr
)
from analysis.plotting import (
    plot_run, plot_multiple_runs, plot_heatmap,
    plot_time_series, plot_scatter, create_summary_figure
)
from analysis.export import (
    export_to_csv, export_to_hdf5, export_to_excel,
    export_metadata, export_for_analysis
)

# Configure matplotlib for inline display
%matplotlib inline
plt.rcParams['figure.figsize'] = (12, 6)
plt.rcParams['font.size'] = 10

print("✅ All imports successful!")
```

## 1. Connecting to DataBroker

DataBroker provides access to all saved experiment runs.

```python
# Connect to catalog
# Try persistent catalog first, fall back to temp if not available
try:
    db = Broker.named('ejfat_gnuradio')
    print("✅ Connected to persistent catalog: ejfat_gnuradio")
except:
    db = Broker.named('temp')
    print("⚠️  Using temporary catalog (data will not persist)")

# Count total runs
try:
    all_runs = list(db.v1())
    print(f"📊 Total runs in catalog: {len(all_runs)}")
except:
    print("📊 No runs found in catalog (run some experiments first!)")
```

## 2. Retrieving Runs

There are multiple ways to retrieve experiment runs from the catalog.

### 2.1 Get Most Recent Run

```python
# Get latest run
try:
    run = db[-1]
    start = run.metadata['start']
    
    print("Latest Run:")
    print(f"  UID: {start['uid'][:16]}...")
    print(f"  Scan ID: {start['scan_id']}")
    print(f"  Plan: {start['plan_name']}")
    print(f"  Operator: {start.get('operator', 'N/A')}")
    print(f"  Purpose: {start.get('purpose', 'N/A')}")
    print(f"  Time: {datetime.fromtimestamp(start['time'])}")
    
except IndexError:
    print("❌ No runs found. Run an experiment first!")
    run = None
```

### 2.2 Search by Criteria

```python
# Search examples
if run is not None:
    # Search by plan name
    plan_name = start['plan_name']
    matching_runs = list(db.search(plan_name=plan_name))
    print(f"Runs with plan '{plan_name}': {len(matching_runs)}")
    
    # Search by operator (if available)
    if 'operator' in start:
        operator_runs = list(db.search(operator=start['operator']))
        print(f"Runs by operator '{start['operator']}': {len(operator_runs)}")
    
    # Search by date
    today = datetime.now().strftime('%Y-%m-%d')
    todays_runs = list(db.search(since=today))
    print(f"Runs today ({today}): {len(todays_runs)}")
```

### 2.3 List All Available Runs

```python
# List all runs with key metadata
if run is not None:
    print("\nAvailable Runs:")
    print("=" * 100)
    print(f"{'Scan ID':<10} {'UID':<10} {'Plan':<25} {'Operator':<15} {'Purpose':<30}")
    print("=" * 100)
    
    for r in all_runs[-10:][::-1]:  # Show last 10, newest first
        s = r.metadata['start']
        print(f"{s['scan_id']:<10} "
              f"{s['uid'][:8]:<10} "
              f"{s['plan_name']:<25} "
              f"{s.get('operator', 'N/A'):<15} "
              f"{s.get('purpose', 'N/A')[:28]:<30}")
    
    print("=" * 100)
```

## 3. Accessing Data

Once you have a run, you can access its data as a Pandas DataFrame.

```python
if run is not None:
    # Get data as DataFrame
    table = run.table()
    
    print(f"Data Shape: {table.shape[0]} rows × {table.shape[1]} columns")
    print(f"\nColumns: {list(table.columns)}")
    print(f"\nFirst 5 rows:")
    display(table.head())
    
    print(f"\nData Types:")
    display(table.dtypes)
else:
    print("No run available for analysis")
```

### 3.1 Basic DataFrame Operations

```python
if run is not None:
    # Statistical summary
    print("Statistical Summary:")
    display(table.describe())
    
    # Access specific columns
    numeric_cols = table.select_dtypes(include=[np.number]).columns.tolist()
    if len(numeric_cols) >= 2:
        col1, col2 = numeric_cols[:2]
        print(f"\nMean {col1}: {table[col1].mean():.4f}")
        print(f"Mean {col2}: {table[col2].mean():.4f}")
        print(f"Max {col2}: {table[col2].max():.4f}")
        print(f"Min {col2}: {table[col2].min():.4f}")
```

## 4. Statistical Analysis

Perform comprehensive statistical analysis on your data.

### 4.1 Comprehensive Statistics

```python
if run is not None and len(numeric_cols) >= 2:
    signal_key = numeric_cols[1]  # Analyze second numeric column
    
    # Compute comprehensive statistics
    stats = analyze_run(run, signal_key)
    
    print(f"Comprehensive Statistics for '{signal_key}':")
    print("=" * 60)
    print(f"  Count:          {len(table[signal_key])}")
    print(f"  Mean:           {stats['mean']:.6f}")
    print(f"  Std Dev:        {stats['std']:.6f}")
    print(f"  Min:            {stats['min']:.6f}")
    print(f"  Max:            {stats['max']:.6f}")
    print(f"  Median:         {stats['median']:.6f}")
    print(f"  Peak-to-Peak:   {stats['peak_to_peak']:.6f}")
    print(f"  Variance:       {stats['var']:.6f}")
    print(f"  RMS:            {stats['rms']:.6f}")
    print(f"  Skewness:       {stats['skewness']:.6f}")
    print(f"  Kurtosis:       {stats['kurtosis']:.6f}")
    print(f"  25th percentile: {stats['q25']:.6f}")
    print(f"  75th percentile: {stats['q75']:.6f}")
    print(f"  IQR:            {stats['iqr']:.6f}")
    print("=" * 60)
```

### 4.2 Peak Detection

```python
if run is not None and len(numeric_cols) >= 2:
    x_key = numeric_cols[0]
    y_key = numeric_cols[1]
    
    # Find peaks
    peaks = find_peaks(run, x_key, y_key, prominence=0.5)
    
    print(f"\nPeak Detection Results:")
    print("=" * 60)
    print(f"  Number of peaks found: {peaks['num_peaks']}")
    
    if peaks['num_peaks'] > 0:
        print(f"\n  Peak positions ({x_key}):")
        for i, (pos, val) in enumerate(zip(peaks['peak_positions'], peaks['peak_values']), 1):
            print(f"    Peak {i}: {pos:.4f} → {y_key} = {val:.4f}")
        
        # Find strongest peak
        max_idx = np.argmax(peaks['peak_values'])
        print(f"\n  Strongest peak:")
        print(f"    Position: {peaks['peak_positions'][max_idx]:.4f}")
        print(f"    Value: {peaks['peak_values'][max_idx]:.4f}")
    print("=" * 60)
```

### 4.3 Outlier Detection

```python
if run is not None and len(numeric_cols) >= 2:
    # Detect outliers using IQR method
    outliers = detect_outliers(run, y_key, method='iqr', threshold=1.5)
    
    print(f"\nOutlier Detection for '{y_key}':")
    print("=" * 60)
    print(f"  Method: {outliers['method'].upper()}")
    print(f"  Threshold: {outliers['threshold']}")
    print(f"  Number of outliers: {outliers['num_outliers']}")
    print(f"  Percentage: {outliers['outlier_fraction']*100:.2f}%")
    
    if outliers['num_outliers'] > 0:
        print(f"\n  Outlier values:")
        for idx, val in zip(outliers['outlier_indices'][:5], outliers['outlier_values'][:5]):
            print(f"    Index {idx}: {val:.4f}")
        if outliers['num_outliers'] > 5:
            print(f"    ... and {outliers['num_outliers']-5} more")
    print("=" * 60)
```

### 4.4 Correlation Analysis

```python
if run is not None and len(numeric_cols) >= 2:
    # Compute correlation
    corr = compute_correlation(run, x_key, y_key)
    
    print(f"\nCorrelation Analysis: {x_key} vs {y_key}")
    print("=" * 60)
    print(f"  Pearson correlation coefficient: {corr['pearson_r']:.4f}")
    print(f"  Pearson p-value: {corr['pearson_p']:.4e}")
    print(f"  Spearman correlation coefficient: {corr['spearman_r']:.4f}")
    print(f"  Spearman p-value: {corr['spearman_p']:.4e}")
    
    # Interpretation
    r = abs(corr['pearson_r'])
    if r > 0.9:
        strength = "Very Strong"
    elif r > 0.7:
        strength = "Strong"
    elif r > 0.5:
        strength = "Moderate"
    elif r > 0.3:
        strength = "Weak"
    else:
        strength = "Very Weak"
    
    print(f"\n  Interpretation: {strength} correlation")
    print("=" * 60)
```

## 5. Data Visualization

Create various types of plots to visualize your data.

### 5.1 Basic Line Plot

```python
if run is not None and len(numeric_cols) >= 2:
    fig, ax = plot_run(run, x_key, y_key)
    plt.show()
```

### 5.2 Scatter Plot with Statistics

```python
if run is not None and len(numeric_cols) >= 2:
    fig, ax = plt.subplots(figsize=(12, 6))
    
    # Scatter plot
    ax.scatter(table[x_key], table[y_key], alpha=0.6, s=50)
    
    # Add mean line
    mean_val = table[y_key].mean()
    ax.axhline(mean_val, color='red', linestyle='--', label=f'Mean = {mean_val:.4f}')
    
    # Add ±1 std lines
    std_val = table[y_key].std()
    ax.axhline(mean_val + std_val, color='orange', linestyle=':', 
               label=f'+1 std = {mean_val + std_val:.4f}')
    ax.axhline(mean_val - std_val, color='orange', linestyle=':', 
               label=f'-1 std = {mean_val - std_val:.4f}')
    
    ax.set_xlabel(x_key)
    ax.set_ylabel(y_key)
    ax.set_title(f'{y_key} vs {x_key} with Statistics')
    ax.legend()
    ax.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.show()
```

### 5.3 Histogram with Statistics

```python
if run is not None and len(numeric_cols) >= 2:
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))
    
    # Histogram
    ax1.hist(table[y_key], bins=30, alpha=0.7, edgecolor='black')
    ax1.axvline(table[y_key].mean(), color='red', linestyle='--', 
                linewidth=2, label='Mean')
    ax1.axvline(table[y_key].median(), color='green', linestyle='--', 
                linewidth=2, label='Median')
    ax1.set_xlabel(y_key)
    ax1.set_ylabel('Frequency')
    ax1.set_title(f'Distribution of {y_key}')
    ax1.legend()
    ax1.grid(True, alpha=0.3)
    
    # Box plot
    ax2.boxplot(table[y_key], vert=True)
    ax2.set_ylabel(y_key)
    ax2.set_title(f'Box Plot of {y_key}')
    ax2.grid(True, alpha=0.3, axis='y')
    
    plt.tight_layout()
    plt.show()
```

### 5.4 Summary Figure

```python
if run is not None:
    fig = create_summary_figure(run)
    plt.show()
```

## 6. Comparing Multiple Runs

Analyze and compare data from multiple experiment runs.

```python
if run is not None:
    # Get runs with same plan
    plan_name = start['plan_name']
    similar_runs = list(db.search(plan_name=plan_name))[-5:]  # Last 5
    
    if len(similar_runs) > 1:
        print(f"Comparing {len(similar_runs)} runs with plan '{plan_name}'\n")
        
        # Statistical comparison
        if len(numeric_cols) >= 2:
            comparison = compare_runs(similar_runs, y_key)
            
            print("\nComparison Statistics:")
            display(comparison[['scan_id', 'uid', 'mean', 'std', 'min', 'max', 'median']])
            
            # Plot comparison
            fig, ax = plot_multiple_runs(similar_runs, x_key, y_key,
                                         title=f'Comparison of {plan_name} Runs')
            plt.show()
    else:
        print(f"Only one run found with plan '{plan_name}'")
```

## 7. Data Export

Export your data in various formats for further analysis.

### 7.1 Export to CSV

```python
if run is not None:
    # Export to CSV with metadata
    csv_path = export_to_csv(run, '../data/exports/tutorial_export.csv')
    
    # Preview the file
    print("\nFirst 10 lines of exported CSV:")
    with open(csv_path, 'r') as f:
        for i, line in enumerate(f):
            if i >= 10:
                break
            print(line.rstrip())
```

### 7.2 Export to HDF5

```python
if run is not None:
    # Export to HDF5 with compression
    h5_path = export_to_hdf5(run, '../data/exports/tutorial_export.h5', compression=True)
    
    # Read it back to verify
    df_read = pd.read_hdf(h5_path, 'data')
    print(f"\nRead back {len(df_read)} rows from HDF5")
    print(f"Columns: {list(df_read.columns)}")
```

### 7.3 Export Metadata

```python
if run is not None:
    # Export metadata as JSON
    json_path = export_metadata(run, '../data/exports/tutorial_metadata.json')
    
    # Load and display
    import json
    with open(json_path, 'r') as f:
        metadata = json.load(f)
    
    print("\nExported Metadata:")
    for key in ['uid', 'scan_id', 'plan_name', 'operator', 'purpose', 'time']:
        if key in metadata:
            print(f"  {key}: {metadata[key]}")
```

### 7.4 Multi-Format Export

```python
if run is not None:
    # Export in multiple formats at once
    exported = export_for_analysis(
        run,
        output_dir='../data/exports/tutorial_multiformat',
        formats=['csv', 'hdf5', 'json']
    )
    
    print("\nExported files:")
    for fmt, path in exported.items():
        print(f"  {fmt}: {path.name}")
```

## 8. Advanced Analysis Example

Putting it all together: comprehensive analysis of a run.

```python
if run is not None and len(numeric_cols) >= 2:
    print("=" * 80)
    print("COMPREHENSIVE ANALYSIS REPORT")
    print("=" * 80)
    
    # Metadata
    print(f"\n1. EXPERIMENT METADATA")
    print("-" * 80)
    print(f"   Run UID: {start['uid'][:16]}...")
    print(f"   Scan ID: {start['scan_id']}")
    print(f"   Plan: {start['plan_name']}")
    print(f"   Operator: {start.get('operator', 'N/A')}")
    print(f"   Purpose: {start.get('purpose', 'N/A')}")
    print(f"   Time: {datetime.fromtimestamp(start['time'])}")
    
    # Data summary
    print(f"\n2. DATA SUMMARY")
    print("-" * 80)
    print(f"   Total points: {len(table)}")
    print(f"   Columns: {list(table.columns)}")
    print(f"   X variable: {x_key}")
    print(f"   Y variable: {y_key}")
    
    # Statistics
    print(f"\n3. STATISTICAL ANALYSIS ({y_key})")
    print("-" * 80)
    stats = analyze_run(run, y_key)
    print(f"   Mean ± Std: {stats['mean']:.4f} ± {stats['std']:.4f}")
    print(f"   Range: [{stats['min']:.4f}, {stats['max']:.4f}]")
    print(f"   Median: {stats['median']:.4f}")
    print(f"   IQR: {stats['iqr']:.4f}")
    print(f"   Skewness: {stats['skewness']:.4f}")
    print(f"   Kurtosis: {stats['kurtosis']:.4f}")
    
    # Peaks
    print(f"\n4. PEAK ANALYSIS")
    print("-" * 80)
    peaks = find_peaks(run, x_key, y_key, prominence=0.5)
    print(f"   Number of peaks: {peaks['num_peaks']}")
    if peaks['num_peaks'] > 0:
        max_idx = np.argmax(peaks['peak_values'])
        print(f"   Highest peak: {peaks['peak_values'][max_idx]:.4f} at {x_key}={peaks['peak_positions'][max_idx]:.4f}")
    
    # Outliers
    print(f"\n5. OUTLIER DETECTION")
    print("-" * 80)
    outliers = detect_outliers(run, y_key, method='iqr', threshold=1.5)
    print(f"   Outliers found: {outliers['num_outliers']} ({outliers['outlier_fraction']*100:.2f}%)")
    
    # Correlation
    print(f"\n6. CORRELATION ANALYSIS")
    print("-" * 80)
    corr = compute_correlation(run, x_key, y_key)
    print(f"   Pearson r: {corr['pearson_r']:.4f} (p={corr['pearson_p']:.2e})")
    print(f"   Spearman r: {corr['spearman_r']:.4f} (p={corr['spearman_p']:.2e})")
    
    print("\n" + "=" * 80)
    print("ANALYSIS COMPLETE")
    print("=" * 80)
```

## 9. Custom Analysis

Space for your own custom analysis code.

```python
# Your custom analysis here
# Example: Time-based analysis

if run is not None and 'time' in table.columns:
    # Convert to relative time
    table['relative_time'] = table['time'] - table['time'].iloc[0]
    
    # Plot signal over time
    if len(numeric_cols) >= 2:
        fig, ax = plt.subplots(figsize=(12, 5))
        ax.plot(table['relative_time'], table[y_key], '-', linewidth=1.5)
        ax.set_xlabel('Time (s)')
        ax.set_ylabel(y_key)
        ax.set_title(f'{y_key} vs Time')
        ax.grid(True, alpha=0.3)
        plt.tight_layout()
        plt.show()
        
        # Compute rate of change
        rate = np.diff(table[y_key]) / np.diff(table['relative_time'])
        print(f"\nRate of change statistics:")
        print(f"  Mean rate: {np.mean(rate):.4e}")
        print(f"  Max rate: {np.max(rate):.4e}")
        print(f"  Min rate: {np.min(rate):.4e}")
```
