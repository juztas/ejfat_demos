# Quick Start: Data Analysis

**Get started analyzing your experiment data in 5 minutes!**

This notebook shows the essentials for accessing and analyzing your Bluesky experiment data.

---

## 1. Setup (Run Once)

````python
# Imports
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent))

import numpy as np
import matplotlib.pyplot as plt
from databroker import Broker
from analysis.statistics import analyze_run, find_peaks
from analysis.plotting import plot_run
from analysis.export import export_to_csv

%matplotlib inline
plt.rcParams['figure.figsize'] = (10, 6)

print("✅ Ready!")
````

## 2. Load Your Data

````python
# Connect to catalog
try:
    db = Broker.named('ejfat_gnuradio')
    print("✅ Connected to catalog")
except:
    db = Broker.named('temp')
    print("⚠️  Using temporary catalog")

# Get latest run
try:
    run = db[-1]
    table = run.table()
    start = run.metadata['start']
    
    print(f"\n📊 Latest Run:")
    print(f"   Plan: {start['plan_name']}")
    print(f"   Points: {len(table)}")
    print(f"   Columns: {list(table.columns)}")
except:
    print("❌ No runs found. Run an experiment first!")
    run = None
````

## 3. View Your Data

````python
if run is not None:
    # Show first few rows
    display(table.head(10))
    
    # Quick statistics
    print("\nQuick Statistics:")
    display(table.describe())
````

## 4. Plot Your Data

````python
if run is not None:
    # Get numeric columns
    numeric_cols = table.select_dtypes(include=[np.number]).columns.tolist()
    
    if len(numeric_cols) >= 2:
        x_col, y_col = numeric_cols[:2]
        
        # Plot
        fig, ax = plot_run(run, x_col, y_col)
        plt.show()
    else:
        print("Need at least 2 numeric columns to plot")
````

## 5. Analyze Your Data

````python
if run is not None and len(numeric_cols) >= 2:
    # Statistics
    stats = analyze_run(run, y_col)
    
    print(f"📈 Analysis of '{y_col}':")
    print(f"   Mean: {stats['mean']:.4f}")
    print(f"   Std: {stats['std']:.4f}")
    print(f"   Range: [{stats['min']:.4f}, {stats['max']:.4f}]")
    print(f"   Median: {stats['median']:.4f}")
    
    # Find peaks
    peaks = find_peaks(run, x_col, y_col, prominence=0.5)
    print(f"\n🔍 Peaks found: {peaks['num_peaks']}")
    
    if peaks['num_peaks'] > 0:
        max_idx = np.argmax(peaks['peak_values'])
        print(f"   Strongest: {peaks['peak_values'][max_idx]:.4f} "
              f"at {x_col}={peaks['peak_positions'][max_idx]:.4f}")
````

## 6. Export Your Data

````python
if run is not None:
    # Export to CSV
    csv_path = export_to_csv(run, '../data/exports/my_data.csv')
    print(f"✅ Data saved to: {csv_path}")
````

## Done! 🎉

You've just:
1. ✅ Loaded your experiment data
2. ✅ Viewed the data table
3. ✅ Created a plot
4. ✅ Analyzed statistics and peaks
5. ✅ Exported to CSV
