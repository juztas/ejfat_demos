# Retrieving and Analyzing Data

Once data is captured, you can retrieve and analyze it using DataBroker and analysis tools.

## Methods for Data Retrieval

1. **DataBroker (Python)** - Programmatic access
2. **retrieve_data.py (CLI)** - Command-line tool
3. **Direct file reading** - Read JSON Lines files

## Method 1: Using DataBroker

````python
from databroker import Broker

# Connect to catalog
db = Broker.named('temp')

# Get latest run
run = db[-1]

print(f"✓ Retrieved run: {run.metadata['start']['uid'][:8]}")
print(f"  Plan: {run.metadata['start']['plan_name']}")
print(f"  Time: {run.metadata['start']['time']}")
````

### Get Data as DataFrame

````python
# Convert to pandas DataFrame
table = run.table()

print(f"Shape: {table.shape}")
print(f"\nColumns: {list(table.columns)}")
print(f"\nFirst few rows:")
table.head()
````

### Statistical Analysis

````python
import sys
from pathlib import Path

parent_dir = Path.cwd().parent if 'jupyter_book_tutorial' in str(Path.cwd()) else Path.cwd()
if str(parent_dir) not in sys.path:
    sys.path.insert(0, str(parent_dir))

from analysis.statistics import analyze_run, find_peaks

# Analyze detector data
stats = analyze_run(run, 'det_value')

print("Statistical Analysis:")
print("=" * 50)
for key, value in stats.items():
    print(f"{key:12s}: {value:.4f}")
````

### Visualization

````python
from analysis.plotting import plot_run
import matplotlib.pyplot as plt

fig, ax = plot_run(run, 'motor', 'det_value')
plt.show()
````

### Export Data

````python
from analysis.export import export_to_csv, export_to_hdf5

# Export to CSV
csv_path = parent_dir / 'data' / 'exports' / f"run_{run.metadata['start']['uid'][:8]}.csv"
export_to_csv(run, str(csv_path))
print(f"✓ Exported to CSV: {csv_path}")

# Export to HDF5
h5_path = parent_dir / 'data' / 'exports' / f"run_{run.metadata['start']['uid'][:8]}.h5"
export_to_hdf5(run, str(h5_path))
print(f"✓ Exported to HDF5: {h5_path}")
````

## Method 2: Command-Line Tool

Use `retrieve_data.py` from terminal:

```bash
# List all runs
python retrieve_data.py --list

# View latest run
python retrieve_data.py --latest --all

# Export to CSV
python retrieve_data.py --latest --export csv
```
